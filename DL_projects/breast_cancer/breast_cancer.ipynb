```python
import pandas as pd

import torch
import torch.nn as nn
import torch.optim as optim

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

```

```python
data = pd.read_csv("C:\\Learning_ML\\DL_projects\\breast_cancer\\Breast_Cancer.csv")
data.head(10)
```

```python
print(list(data.columns))
```

```python
for i in ['Race', 'Marital Status', 'T Stage', 'N Stage', '6th Stage', 'differentiate', 'Grade', 'A Stage', 'Estrogen Status', 'Progesterone Status']:
    res = []
    for j in data[i]:
        if j not in res:
            res.append(j)
    print(i,' : ',res,' : ',len(res))
```

```python
X = data.drop(columns='Status')
Y = data['Status']
X.head()
```

```python
label = LabelEncoder()
cat = X.select_dtypes(include='object').columns
for i in cat:
    X[i] = label.fit_transform(X[i])
X.head(10)
```

```python
print(cat)
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(X, Y, random_state=42, shuffle=True, test_size=0.1)
```

```python
scaler = StandardScaler()
xtrain = scaler.fit_transform(xtrain)
xtest = scaler.transform(xtest)
```

```python
le = LabelEncoder()
ytrain = le.fit_transform(ytrain)
ytest = le.fit_transform(ytest)
```

```python
xtrain
```

```python
ytrain
print(type(ytrain))
ytest
print(type(ytest))
```

```python
xtrain = torch.tensor(xtrain, dtype=torch.float32)
xtest = torch.tensor(xtest, dtype=torch.float32)
ytrain = torch.tensor(ytrain, dtype=torch.float32).reshape(-1,1)
ytest = torch.tensor(ytest, dtype=torch.float32).reshape(-1,1)
```

```python
ytest.dtype
```

```python
class NN_Model(nn.Module):
    def __init__(self, parameters):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(parameters, 4),
            nn.ReLU(),
            nn.Linear(4,1)
        )
    
    def forward(self, x):
        return self.network(x)
```

```python
model = NN_Model(xtrain.shape[0])
```

```python
epoch = 200
lr = 0.001
```

```python
optimize = optim.Adam(model.parameters(), lr)
```

```python
for ep in range(epoch):
    output = model(xtrain)
    loss = nn.BCELoss(output, ytrain)
    optimize.zero_grad
    loss.backward()
    optimize.step()
    if ep%25==0:
        print(f"For {ep} : {loss.items():.4f}")

```
