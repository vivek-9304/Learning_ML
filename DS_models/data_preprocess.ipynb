```python
from sklearn.preprocessing import (
    StandardScaler, MinMaxScaler, RobustScaler, MaxAbsScaler, Normalizer,
    LabelEncoder, OneHotEncoder, OrdinalEncoder, LabelBinarizer,
    PolynomialFeatures, KBinsDiscretizer, 
    FunctionTransformer, PowerTransformer, QuantileTransformer, SplineTransformer, Binarizer, MultiLabelBinarizer
)
import numpy as np
```

```python
# SCALING
```

```python
'''
Standard Scaler:

formula: z = (x - mean) / std

Works well for:
- Logistic Regression
- Linear Regression
- SVM
- Neural Networks
- KNN
- PCA

When NOT ideal
- If data contains many outliers.

'''

X = np.array([[10], [20], [30], [40]])
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)

X_original = scaler.inverse_transform(X_scaled) # To reverse the scaling and get back the original data
print(X_original)
```

```python
'''
min-max scaler:

formula: z = (x - min) / (max - min)

Works well for:
Exact range of data is known and outliers are not present.
- Neural Networks
- Image Processing
- Deep Learning

When NOT ideal
- Very sensitive to outliers. Outliers can skew the min and max values, leading to distorted scaling of the data.

'''

X = np.array([[10], [20], [30], [40]])
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

```python
'''
Robust Scaler:

formula: z = (x - median) / IQR

Works well for:
- Data with outliers
- Data that is not normally distributed

When NOT ideal
- If data is normally distributed and does not contain outliers, other scalers like Standard Scaler may be more appropriate.

'''

X = np.array([[10], [20], [30], [40], [1000]])
scaler = RobustScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

```python
'''
MaxAbs Scaler:

formula: z = x / max(abs(x))

Works well for:
- Data that is already centered around zero and has a known maximum absolute value.
- Sparse data (e.g., text data represented as word counts)
- Data that is not normally distributed

When NOT ideal
- If data contains outliers, as it can be sensitive to them.
- If data is not centered around zero, as it may not effectively scale the data.

'''

X = np.array([[-10], [-5], [0], [5], [10]])
scaler = MaxAbsScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

```python
'''
normalizer:

formula: z = x / ||x||

Works well for:
- Data that is not normally distributed
- Data that is on different scales
- Text data represented as word counts

When NOT ideal
- If data is already on the same scale, as it may not provide any additional benefit.

'''

X = np.array([
    [3, 4],
    [1, 2],
    [10, 10]
])
normalizer = Normalizer()
X_scaled = normalizer.fit_transform(X)
print(X_scaled)
```

```python
# ENCODING
```

```python
# Label Encoding:

le = LabelEncoder()
data = ["Male", "Female", "Male", "Transgender"]
encoded = le.fit_transform(data)
print(encoded)

# Best for: A pair of categorical values (binary classification) where the order of the values does not matter.
```

```python
# One-Hot Encoding:

X = np.array([["red"], ["blue"], ["green"], ["red"]])
ohe = OneHotEncoder(sparse_output=False)
X_encoded = ohe.fit_transform(X)
print(X_encoded)

# best for: Categorical variables with more than two categories, where the order of the categories does not matter.
```

```python
# Ordinal Encoding:

X = np.array([
    ["small"],
    ["medium"],
    ["large"],
    ["medium"]
])

oe = OrdinalEncoder(categories=[["small", "medium", "large"]])
X_encoded = oe.fit_transform(X)
print(X_encoded)

# Best for: Categorical variables with more than two categories, where the order of the categories matters.
```

```python
# Label Binarizer:

y = ["yes", "no", "yes", "no"]
lb = LabelBinarizer()
y_encoded = lb.fit_transform(y)
print(y_encoded)
```

```python
'''
Feature Generation:

When to use
- Linear Regression
- Logistic Regression
- Curve fitting

'''

poly = PolynomialFeatures(degree=3)
X = [[1], [2], [3]]
X_poly = poly.fit_transform(X)
print(X_poly)
```

```python
'''
KBins Discretizer:

When to use
- Credit scoring
- Risk analysis
- Age grouping

'''
X = [[18], [25], [40], [60], [80]]
kbd = KBinsDiscretizer(
    n_bins=3,
    encode="ordinal",
    strategy="uniform"
)
print(kbd.fit_transform(X))
```

```python
'''
Function Transformer:

When useful: 
- Logarithmic transformation

'''
transformer = FunctionTransformer(np.log1p)
X = [[1], [10], [100]]
print(transformer.fit_transform(X))
```

```python
'''
Power Transformer:

When useful:
- When the data has a skewed distribution
- When the data has a non-normal distribution

When to use:
- Linear Regression
- Logistic Regression
- Statistical models

'''
pt = PowerTransformer()
X_transformed = pt.fit_transform(X)
print(X_transformed)
```

```python
'''
Quantile Transformer:

When to use:
- Extreme outliers
- Non-normal distributions
'''

qt = QuantileTransformer()
X_new = qt.fit_transform(X)
```

```python
'''
Spline Transformer:

Used in
- Time series
- Forecasting
- Smooth nonlinear relationships

'''
spline = SplineTransformer(n_knots=4)
X_new = spline.fit_transform(X)
```

```python
'''
Binarizer:

When to use
- Presence/absence
- Boolean features
- Rule-based systems
'''

b = Binarizer(threshold=5)
print(b.fit_transform([[2], [5], [8], [12]]))
```

```python
'''
MultiLabel Binarizer:

Used in:
- Tags
- Genres
- Recommendations
'''

mlb = MultiLabelBinarizer()
data = [
    ["Action", "Comedy"],
    ["Drama"],
    ["Action", "Drama"]
]
print(mlb.fit_transform(data))
```
