```python
df = pd.DataFrame({'a': [1, np.nan, 3], 'b': [4, 5, np.nan]})

df.isnull()          # True where NaN
df.isnull().sum()    # count NaNs per column

df.dropna()          # drop rows with any NaN
df.dropna(axis=1)   # drop columns with any NaN
df.dropna(thresh=2) # keep rows with ≥2 non-NaN

df.fillna(0)                     # fill all NaN with 0
df['a'].fillna(df['a'].mean())   # fill with column mean
df.ffill()                       # forward-fill
df.bfill()                       # backward-fill
```

```python
df.duplicated()                     # boolean mask
df.duplicated().sum()              # count duplicates

df.drop_duplicates()               # drop all dupes
df.drop_duplicates(subset=['name']) # based on column
df.drop_duplicates(keep='last')   # keep last occurrence
```

```python
# Rename columns
df.rename(columns={'name': 'full_name', 'age': 'years'})
# Drop column(s)
df.drop(columns=['score'])
df.drop(columns='score', inplace=True)
# Drop row(s) by index
df.drop(index=[0, 2])
```

```python
df['age'] = df['age'].astype('float')
df['score'] = df['score'].astype('int')
df['name'] = df['name'].astype('category')  # saves memory

# Convert to datetime
df['date'] = pd.to_datetime(df['date'])
df['id']   = pd.to_numeric(df['id'], errors='coerce')
```

```python
df.sort_values('score')                          # ascending
df.sort_values('score', ascending=False)         # descending
df.sort_values(['age', 'score'], ascending=[True, False])

df.sort_index()           # sort by row index
df.sort_index(axis=1)     # sort columns alphabetically
```

```python
# Apply to a single column (Series)
df['score'].apply(lambda x: 'pass' if x >= 80 else 'fail')

# Apply to each row (axis=1)
df.apply(lambda row: row['age'] * row['score'], axis=1)

# Apply to each column (axis=0, default)
df[['age', 'score']].apply('mean')

# Named function
def normalize(x):
    return (x - x.min()) / (x.max() - x.min())
df[['score']].apply(normalize)
```

```python
# Map values using a dict
df['grade'] = df['score'].map({
    88.5: 'B+', 92.0: 'A-', 78.3: 'C+'
})

# Map using a function
df['name_upper'] = df['name'].map(str.upper)

# Replace specific values
df['score'].replace(78.3, 79.0)
df.replace({'name': {'Bob': 'Robert'}})
```

```python
df1 = pd.DataFrame({'id':[1,2], 'val':['a','b']})
df2 = pd.DataFrame({'id':[2,3], 'extra':['x','y']})

# Merge (inner by default)
pd.merge(df1, df2, on='id')
pd.merge(df1, df2, on='id', how='left')   # left join
pd.merge(df1, df2, on='id', how='outer')  # full outer

# Concat (stack vertically)
pd.concat([df1, df2], ignore_index=True)

# Concat horizontally (side by side)
pd.concat([df1, df2], axis=1)
```

```python
# iterrows → (index, Series)
for idx, row in df.iterrows():
    print(idx, row['name'], row['score'])

# itertuples → namedtuple (faster than iterrows)
for row in df.itertuples():
    print(row.Index, row.name, row.score)

# Prefer vectorized operations instead:
df['score'] * 2        # much faster than looping
```

```python
df = pd.DataFrame({
    'dept': ['eng', 'eng', 'hr', 'hr'],
    'salary': [90, 95, 60, 65]
})

df.groupby('dept')['salary'].mean()

df.groupby('dept').agg({'salary': ['mean', 'max', 'count']})
```

```python
df = pd.DataFrame({
    'region': ['N', 'N', 'S', 'S'],
    'product': ['A', 'B', 'A', 'B'],
    'sales': [100, 200, 150, 250]
})

df.pivot_table(
    values='sales',
    index='region',
    columns='product',
    aggfunc='sum',
    fill_value=0
)
```

```python
# agg: one row per group
df.groupby('dept')['salary'].agg(['mean', 'std', 'count'])

# transform: broadcasts back to original shape
df['salary_zscore'] = df.groupby('dept')['salary'].transform(
    lambda x: (x - x.mean()) / x.std()
)
# Named aggregation (pandas ≥ 0.25)
df.groupby('dept').agg(
    avg_sal=('salary', 'mean'),
    max_sal=('salary', 'max')
)
```
