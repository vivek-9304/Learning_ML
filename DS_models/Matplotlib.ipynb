```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
from mpl_toolkits.mplot3d import Axes3D  # registers '3d' projection
import numpy as np
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title('Sine Wave')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(x, y, label='sin(x)')
ax.plot(x, np.cos(x), label='cos(x)')

ax.set_title('Trig Functions')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()

plt.show()
```

```python
# Single figure
fig = plt.figure(figsize=(10, 6), dpi=100)

# Figure + single Axes (most common)
fig, ax = plt.subplots(figsize=(8, 5))

# Figure + grid of Axes
fig, axes = plt.subplots(2, 3, figsize=(12, 8))
# axes is a 2D array of Axes objects: axes[0,0], axes[1,2], etc.

# Shared axes
fig, axes = plt.subplots(2, 1, sharex=True, sharey=True)
```

```python
fig, ax = plt.subplots()

ax.plot(x, y, color='blue', linewidth=2,
        linestyle='--', marker='o', markersize=4,
        label='data', alpha=0.8)

# Short-hand format string: 'color-marker-linestyle'
ax.plot(x, y, 'r--o')   # red dashed line with circle markers
ax.plot(x, y, 'g-')     # green solid line
ax.plot(x, y, 'bs')     # blue squares, no line
```

```python
x = np.random.rand(50)
y = np.random.rand(50)
sizes = np.random.rand(50) * 300
colors = np.random.rand(50)

fig, ax = plt.subplots()
sc = ax.scatter(x, y, s=sizes, c=colors, cmap='viridis',
                 alpha=0.6, edgecolors='black')

plt.colorbar(sc, label='color value')
ax.set_title('Scatter Plot')
```

```python
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 12, 37]

fig, ax = plt.subplots()
ax.bar(categories, values, color='steelblue', edgecolor='black')

# Horizontal
ax.barh(categories, values)

# Grouped bars
x = np.arange(len(categories))
width = 0.35
fig, ax = plt.subplots()
ax.bar(x - width/2, values, width, label='2024')
ax.bar(x + width/2, [30,40,20,25], width, label='2025')
ax.set_xticks(x)
ax.set_xticklabels(categories)
ax.legend()
```

```python
data = np.random.randn(1000)

fig, ax = plt.subplots()
ax.hist(data, bins=30, color='skyblue',
        edgecolor='black', alpha=0.7)

ax.set_title('Distribution')
ax.set_xlabel('Value')
ax.set_ylabel('Frequency')

# Multiple overlapping histograms
ax.hist(np.random.randn(1000), bins=30, alpha=0.5, label='A')
ax.hist(np.random.randn(1000)+1, bins=30, alpha=0.5, label='B')
ax.legend()

# density=True normalizes to a probability density
ax.hist(data, bins=30, density=True)
```

```python
data = [np.random.randn(100), np.random.randn(100)+1, np.random.randn(100)*2]

fig, ax = plt.subplots()
ax.boxplot(data, labels=['A', 'B', 'C'],
           patch_artist=True, showmeans=True)

# Violin plot — shows full distribution shape
fig, ax = plt.subplots()
ax.violinplot(data, showmedians=True)
ax.set_xticks([1,2,3])
ax.set_xticklabels(['A', 'B', 'C'])
```

```python
sizes = [35, 25, 20, 20]
labels = ['Product A', 'Product B', 'Product C', 'Product D']
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']
explode = (0.05, 0, 0, 0)  # slightly pull out 1st slice

fig, ax = plt.subplots()
ax.pie(sizes, labels=labels, colors=colors, explode=explode,
       autopct='%1.1f%%', startangle=90, shadow=True)
ax.axis('equal')  # keep circle shape
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
y_upper = y + 0.2
y_lower = y - 0.2

fig, ax = plt.subplots()
ax.plot(x, y, color='blue', label='mean')
ax.fill_between(x, y_lower, y_upper, alpha=0.3,
                color='blue', label='confidence band')

# Conditional fill (shade only where condition true)
ax.fill_between(x, y, 0, where=(y > 0), color='green', alpha=0.3)
ax.fill_between(x, y, 0, where=(y <= 0), color='red', alpha=0.3)
```

```python
data = np.random.rand(10, 10)

fig, ax = plt.subplots()
im = ax.imshow(data, cmap='coolwarm', interpolation='nearest')
plt.colorbar(im, ax=ax)

# Annotated heatmap (e.g. correlation matrix)
fig, ax = plt.subplots()
im = ax.imshow(data, cmap='viridis')
for i in range(10):
    for j in range(10):
        ax.text(j, i, f'{data[i,j]:.1f}', ha='center', va='center', color='w')
```

```python
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

fig, ax = plt.subplots()

# Line contours
cs = ax.contour(X, Y, Z, levels=10, colors='black')
ax.clabel(cs, inline=True, fontsize=8)

# Filled contours
fig, ax = plt.subplots()
cf = ax.contourf(X, Y, Z, levels=20, cmap='RdBu')
plt.colorbar(cf)
```

```python

fig = plt.figure(figsize=(10, 8))
gs = GridSpec(3, 3, figure=fig)

ax1 = fig.add_subplot(gs[0, :])      # top row, full width
ax2 = fig.add_subplot(gs[1, :-1])   # middle row, first 2 cols
ax3 = fig.add_subplot(gs[1:, -1])   # last col, spans 2 rows
ax4 = fig.add_subplot(gs[-1, 0])     # bottom-left
ax5 = fig.add_subplot(gs[-1, -2])    # bottom-middle

ax1.set_title('ax1: full width')
plt.tight_layout()
```

```python
fig, ax1 = plt.subplots()

ax1.plot(x, np.sin(x), color='blue')
ax1.set_xlabel('x')
ax1.set_ylabel('sin(x)', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')

# Second y-axis sharing the same x-axis
ax2 = ax1.twinx()
ax2.plot(x, np.exp(x), color='red')
ax2.set_ylabel('exp(x)', color='red')
ax2.tick_params(axis='y', labelcolor='red')

fig.tight_layout()
```

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# 3D surface plot
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

ax.plot_surface(X, Y, Z, cmap='viridis')

# 3D scatter
ax.scatter(np.random.rand(50), np.random.rand(50), np.random.rand(50))

ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
```

```python
# Get current figure / axes
fig = plt.gcf()   # get current figure
ax = plt.gca()    # get current axes

# Clear without closing (reuse figure)
plt.cla()   # clear current axes
plt.clf()   # clear current figure

# Close figures (free memory — important in loops!)
plt.close()       # close current figure
plt.close(fig)    # close specific figure
plt.close('all')  # close all figures

# Adjust spacing
plt.tight_layout()
plt.subplots_adjust(hspace=0.4, wspace=0.3)
```
