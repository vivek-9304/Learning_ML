```python
from sklearn.feature_selection import (
    SelectKBest, VarianceThreshold, SelectPercentile, RFE, RFECV, SelectFromModel, 
    chi2, f_classif, f_regression, mutual_info_classif, mutual_info_regression
)
from sklearn.linear_model import LogisticRegression
import pandas as pd
```

```python
'''
VarianceThrehold :
Used when : First cleaning step, constant variables

'''
df = pd.DataFrame({
    "Age": [18,19,18,20],
    "StudyHours": [5,2,7,4],
    "Country": [1,1,1,1],  # Same value everywhere
    "Marks": [80, 50, 90, 60]
})
X = df.drop("Marks", axis=1)
y = df["Marks"]
print(df)

selector = VarianceThreshold(
    threshold=0
)
X_new = selector.fit_transform(X)
print(X_new)
```

```python
df = pd.DataFrame({
    "Age": [18,19,18,20,21],
    "StudyHours": [5,2,7,4,8],
    "MobileUsage": [8,10,2,7,1],
    "Pass": [1,0,1,0,1]
})
X = df.drop("Pass", axis=1)
y = df["Pass"]

selector = SelectKBest(
    score_func=f_classif,
    k=10
)
X_new = selector.fit_transform(X, y)
print(X_new)
```

```python
model = LogisticRegression()
selector = RFE(
    estimator=model,
    n_features_to_select=2
)
X_new = selector.fit_transform(X, y)
print(X_new)

selected_features = X.columns[
    selector.support_
]
print(selected_features)
```
