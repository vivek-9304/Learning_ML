```python
# Removing duplicates elements
ls = [1,2,3,3,3,4,5,6,6,6,7,7,7,9]
result = []
for i in ls:
    if i not in result:
        result.append(i)
print(result)
```

```python
# enumerate the data of list
lst = [10, 20, 30]
for i, val in enumerate(lst):
    print(i, val)
```

```python
# form a dict with two list
l1 = ['0','2','3','4']
l2 = [5,6,7,8]
d = dict(zip(l1,l2))
print(d)
```

```python
# finding common element
a = [1,2,3,4]
b = [3,4,5,6]
result = []
for i in a:
    for j in b:
        if i==j:
            result.append(i)
print(result)
```

```python
# frequency of objects
data = "apple banana apple mango banana apple"
l = data.split()
result = {}
for i in l:
    result[i] = result.get(i,0)+1
print(result)
```

```python
# sum of tuple inside list
data = [(1,2), (3,4), (5,6)]
for i,j in data:
    print(i+j)
```

```python
# check palindrome
data = input().strip()
print(data)
str=""
for i in reversed(data):
    str = str+i
if str==data:
    print("Palindrome")
else:
    print("NO")
```

```python
b = [3, 4, 5, 6]
l = [1, 2, 2, 3, 4, 4]
res = l+b
res.sort()
print(res)
n = []
for i in res:
    if i not in n:
        n.append(i)
print(n)
```
