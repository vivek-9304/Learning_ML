```python
class Student():
    def __init__(self, name, age, branch):
        self.name = name
        self.age = age
        self.branch = branch
    def introduce(self):
        print(f"Hi, I am {self.name}")
        print(f"I am {self.age} years old")
        print(f"I study {self.branch}")
        
stu = Student("Vivek",21,"AIML")
stu.introduce()
```

```python
class Student():
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks
    def introduce(self):
        print(f"Hi, I am {self.name}")
        print(f"I am {self.age} years old")
        print(f"I scored {self.marks}")
    def is_pass(self):
        return True if self.marks>=40 else False

stu = Student("Vivek",21,28)
stu.introduce()
if stu.is_pass():
    print("Passed") 
else:
    print("Failed")
```

```python
class BankBalance():
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
    def intro(self):
        print(f"Name is {self.name}")
        print(f"Your Balance is {self.balance:.2f}")
    def Deposit(self, deposit):
        self.balance += deposit
    def Withdrawal(self, withdraw):
        if withdraw<=self.balance:
            self.balance -= withdraw
        else:
            print("ERROR: Can't Withdraw sum more than balance")
    def Balance(self):
        print(f"Your Balance is {self.balance:.2f}")
        
bank = BankBalance("Vivek", 12000.00)
bank.intro()
bank.Deposit(4000.00)
bank.Withdrawal(2000.00)
bank.Balance()
```
