```python
student = {
    "name": "Titan",
    "age": 19,
    "course": "AI/ML",
    "marks" : "Fail"
}

# 1. Accessing values
print(student["name"])

# 2. Using get()
print(student.get("age"))

# 3. get() with default value
print(student.get("marks", "Not Found"))

# 4. Adding new key
student["city"] = "Delhi"

# 5. Updating value
student["age"] = 20

# 6. update()
student.update({"college": "DU", "age": 21})

# 7. keys()
print(student.keys())

# 8. values()
print(student.values())

# 9. items()
print(student.items())

# 10. Loop through dictionary
for key, value in student.items():
    print(key, ":", value)

# 11. Check key exists
if "name" in student:
    print("Name exists")

# 12. pop()
removed = student.pop("city")
print("Removed:", removed)

# 13. copy()
copy_dict = student.copy()

# 14. len()
print(len(student))

# 15. clear()
temp = {"a": 1, "b": 2}
temp.clear()
print(temp)

# 16. Nested dictionary
students = {
    "s1": {"name": "Aman", "marks": 90},
    "s2": {"name": "Rahul", "marks": 85}
}

print(students["s1"]["marks"])
```

```python
# Problem 1 — Frequency Counter
a = "banana"
d = {}
for i in a:
    d[i] = d.get(i,0)+1
print(d)
```

```python
# Problem 2 — Count Words
a = "hi hello hi titan"
l = list(a.split(" "))
res = {}
for i in l:
    res[i] = res.get(i,0)+1
print(res)
```

```python
# Problem 3 — Find Highest Marks
students = {
    "Aman": 78,
    "Rahul": 92,
    "Titan": 95,
    "Neha": 88
}
max = students["Aman"]
name = "Aman"
for key,value in students.items():
    if value>max:
        max = value
        name = key
print(name ,":",max)
```

```python
# Problem 4 — Reverse Dictionary
d = {"a":1, "b":2, "c":3}
d_out = {}
for key, value in d.items():
    d_out[value] = key
print(d_out)
```

```python
# Problem 5 — Duplicate Finder
lst = [1,2,3,2,5,1,7,7]
res = []
for i in range(0,len(lst)-1):
    for j in range(i+1, len(lst)):
        if lst[i]==lst[j]:
            res.append(lst[i])
print(res)
```
