```python
import csv

path = r"C:\Learning_ML\ML_Projects\Cybersecurity_Agent\data\master_ai_agent_cybersecurity.csv"

with open(path, "r", encoding="utf-8", newline="") as f:
    reader = csv.reader(f)

    expected_cols = None

    for line_no, row in enumerate(reader, start=1):
        if expected_cols is None:
            expected_cols = len(row)
            print(f"Header has {expected_cols} columns")
            continue

        if len(row) != expected_cols:
            print(f"\n❌ Problem at line {line_no}")
            print(f"Expected {expected_cols} columns, found {len(row)}")
            print(row)
```
