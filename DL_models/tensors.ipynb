```python
t = torch.tensor([10, 20, 30, 40, 50])

# Boolean mask
mask = t > 25                   # tensor([F, F, T, T, T])
t[mask]                          # tensor([30, 40, 50])
t[t % 20 == 0]                 # tensor([20, 40])

# torch.where — conditional selection
torch.where(t > 25, t, torch.zeros_like(t))
# tensor([ 0,  0, 30, 40, 50])

# Index tensor (fancy indexing)
idx = torch.tensor([0, 2, 4])
t[idx]                           # tensor([10, 30, 50])

# nonzero: indices of True elements
torch.nonzero(t > 25)          # tensor([[2],[3],[4]])
```

```python
t = torch.arange(12)           # shape (12,)

t.reshape(3, 4)                 # (3, 4)  — copy if needed
t.view(3, 4)                    # (3, 4)  — view (must be contiguous)
t.reshape(3, -1)                # -1 infers dim → (3, 4)
t.reshape(2, 2, 3)              # (2, 2, 3)

t.flatten()                      # always 1D
t.reshape(-1)                   # same as flatten

# Add / remove dimensions
t.unsqueeze(0)                  # (1, 12) — insert dim at position 0
t.unsqueeze(1)                  # (12, 1)
t.unsqueeze(0).squeeze(0)       # remove dim of size 1
```

```python
a = torch.zeros(2, 3)
b = torch.ones(2, 3)

# cat: join along existing axis
torch.cat([a, b], dim=0)       # (4, 3) — stack rows
torch.cat([a, b], dim=1)       # (2, 6) — stack cols

# stack: creates a NEW axis
torch.stack([a, b], dim=0)     # (2, 2, 3)
torch.stack([a, b], dim=1)     # (2, 2, 3) different axis

# split: divide into chunks of given size
t = torch.arange(12).reshape(6, 2)
chunks = torch.split(t, 2, dim=0)   # 3 tensors of shape (2,2)

# chunk: divide into N equal pieces
torch.chunk(t, 3, dim=0)             # 3 tensors of shape (2,2)
```

```python
t = torch.randn(2, 3, 4)

# permute: reorder ALL dimensions
t.permute(0, 2, 1)             # (2, 4, 3)
t.permute(2, 0, 1)             # (4, 2, 3)

# transpose: swap two dimensions
t.transpose(1, 2)              # (2, 4, 3)

# .T shortcut for 2D tensors
m = torch.randn(3, 4)
m.T                             # (4, 3)

# After permute/transpose the tensor may not be contiguous
# Call .contiguous() before .view()
t.permute(2, 0, 1).contiguous().view(-1)
```

```python
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([4.0, 5.0, 6.0])

a + b                 # tensor([5., 7., 9.])
a - b                 # tensor([-3., -3., -3.])
a * b                 # tensor([ 4., 10., 18.])  element-wise
a / b                 # tensor([0.25, 0.40, 0.50])
a ** 2                # tensor([1., 4., 9.])

# Equivalent torch functions
torch.add(a, b)
torch.sub(a, b)
torch.mul(a, b)
torch.div(a, b)

# In-place versions (modify tensor directly, saves memory)
a.add_(b)             # a = a + b  (trailing _ = in-place)
a.mul_(2.0)          # a = a * 2
```

```python
t = torch.tensor([-2.0, 0.0, 1.0, 4.0])

torch.exp(t)           # e^x
torch.log(t)           # natural log (NaN for negatives)
torch.log2(t)
torch.log10(t)
torch.sqrt(t)          # square root
torch.abs(t)           # tensor([2., 0., 1., 4.])
torch.sign(t)          # tensor([-1., 0., 1., 1.])
torch.ceil(t)
torch.floor(t)
torch.round(t)

# clamp: restrict values to [min, max]
torch.clamp(t, min=0.0, max=3.0)   # tensor([0., 0., 1., 3.])

# Trigonometry
torch.sin(t); torch.cos(t); torch.tanh(t)
```

```python
t = torch.tensor([[1.0, 2.0], [3.0, 4.0]])

# Global reductions → scalar
t.sum()              # tensor(10.)
t.prod()             # tensor(24.)
t.mean()             # tensor(2.5)
t.max()              # tensor(4.)
t.min()              # tensor(1.)

# Along a dimension
t.sum(dim=0)         # tensor([4., 6.])  — column sums
t.sum(dim=1)         # tensor([3., 7.])  — row sums

# keepdim=True preserves shape for broadcasting
t.sum(dim=1, keepdim=True)    # tensor([[3.],[7.]])

# max/min return (values, indices) when dim specified
vals, idx = t.max(dim=1)      # max per row + its index
```

```python
# Scalar broadcast
t = torch.ones(3, 4)
t + 10                          # adds 10 to every element

# Row vector broadcast over matrix
row = torch.tensor([[1.0, 2.0, 3.0, 4.0]])  # (1, 4)
mat = torch.ones(3, 4)                          # (3, 4)
(mat + row).shape                                # (3, 4) ✓

# Column vector broadcast
col = torch.tensor([[1.0], [2.0], [3.0]])     # (3, 1)
(mat + col).shape                                # (3, 4) ✓

# Rule: dimensions align from RIGHT, compatible if equal or one is 1
# (3,4) + (4,) → (3,4)  ✓
# (3,4) + (3,) → ERROR  ✗ — use (3,1) instead
```

```python
a = torch.tensor([3, 1, 4, 1, 5])
b = torch.tensor([2, 2, 2, 2, 2])

a == b             # tensor([F, F, T, F, F])
a > b              # element-wise greater than
a >= b
torch.eq(a, b)     # same as ==

torch.any(a > 3)  # tensor(True)
torch.all(a > 0)  # tensor(True)

# Sort
sorted_t, indices = torch.sort(a)            # ascending
sorted_t, indices = torch.sort(a, descending=True)

# Top-k values
vals, idx = torch.topk(a, k=3)             # 3 largest
```

```python
a = torch.randn(3, 4)
b = torch.randn(4, 5)

# 2D matrix multiply
torch.mm(a, b)             # (3, 5)  — only 2D
torch.matmul(a, b)         # (3, 5)  — supports ND
a @ b                      # @ operator — cleanest syntax

# Batched matrix multiply
A = torch.randn(10, 3, 4)  # batch of 10 matrices
B = torch.randn(10, 4, 5)
torch.bmm(A, B)            # (10, 3, 5)

# Dot product (1D tensors)
u = torch.tensor([1.0, 2.0, 3.0])
v = torch.tensor([4.0, 5.0, 6.0])
torch.dot(u, v)            # tensor(32.) — scalar
```

```python
A = torch.tensor([[3.0, 1.0], [1.0, 2.0]])

torch.linalg.det(A)          # determinant: 5.0
torch.linalg.inv(A)          # inverse matrix
torch.linalg.matrix_rank(A)  # rank: 2
torch.linalg.norm(A)         # Frobenius norm
torch.linalg.norm(A, ord=1) # L1 norm

# Eigenvalues and eigenvectors
vals, vecs = torch.linalg.eig(A)

# Singular Value Decomposition
U, S, Vh = torch.linalg.svd(A)

# Solve Ax = b
b = torch.tensor([9.0, 8.0])
x = torch.linalg.solve(A, b)  # solution vector
```

```python
#  Enable gradient tracking
x = torch.tensor([2.0, 3.0], requires_grad=True)

# Build a computation graph
y = x ** 2         # y = x²
z = y.sum()        # z = sum(x²)

# Backpropagate — compute dz/dx
z.backward()

print(x.grad)      # tensor([4., 6.])  — dz/dx = 2x

# Gradients accumulate! Zero them before next pass
x.grad.zero_()

# Disable gradient tracking (e.g. during inference)
with torch.no_grad():
    result = x ** 2   # no graph built, faster

# Or use .detach() to get a copy without grad history
x_no_grad = x.detach()
```

```python
# Most common dtypes
torch.float32   # 32-bit float  — default for floats, most ML
torch.float64   # 64-bit float  — high precision (numpy default)
torch.float16   # 16-bit float  — half precision (GPU training)
torch.bfloat16  # brain float16 — better range than float16
torch.int32     # 32-bit int
torch.int64     # 64-bit int    — default for ints
torch.bool      # boolean

# Create with specific dtype
t = torch.tensor([1, 2, 3], dtype=torch.float32)

# Cast to another dtype
t.float()               # → float32
t.double()              # → float64
t.int()                 # → int32
t.long()                # → int64
t.bool()                # → bool
t.to(torch.float16)     # explicit cast
```
