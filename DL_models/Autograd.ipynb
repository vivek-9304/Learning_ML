```python
# Calculating Grades
```

```python
x = torch.tensor([8.9, 7.8, 8.6])
y = torch.tensor([1.0, 0.0, 1.0])
```

```python
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
```

```python
def z(x, w, b):
    return w*x + b
def y_pred(x,w,b):
    return torch.sigmoid(z(x,w,b))
def loss_fx(y_pred, y):
    ep = 1e-8
    y_pred = torch.clamp(y_pred, ep, 1-ep)
    l =  -(y*torch.log(y_pred) + (1-y)*torch.log(1-y_pred))
    return l.mean()

pred = y_pred(x,w,b)
loss = loss_fx(pred, y)
```

```python
loss.backward()
```

```python
print(w.grad)
print(b.grad)
```

```python
lr = 0.01
w = w - lr*w.grad
print(w)
```
