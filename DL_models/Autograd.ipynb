```python
x = torch.tensor([8.9,7.2, 7.8, 8.6])
y = torch.tensor([1.0,0.0, 0.0, 1.0])
```

```python
w = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)
```

```python
def linear(x, w, b):
    return w * x + b
def predict(x, w, b):
    z = linear(x, w, b)
    return torch.sigmoid(z)
def loss_fn(pred, y):
    eps = 1e-8
    pred = torch.clamp(pred, eps, 1 - eps)
    loss = -(y * torch.log(pred) + (1 - y) * torch.log(1 - pred))
    return loss.mean()
```

```python
lr = 0.05
epochs = 1001
```

```python
for ep in range(epochs):
    pred = predict(x, w, b)
    loss = loss_fn(pred, y)
    if w.grad is not None:
        w.grad.zero_()
        b.grad.zero_()
    loss.backward()
    with torch.no_grad():
        w = w - lr*w.grad
        b = b - lr*b.grad
    w.requires_grad_(True)
    b.requires_grad_(True)
    if(ep%200==0):
        print("end=epoch: ", ep, "loss: ", loss.item(), "w: ", w.item(), "b: ", b.item())
```

```python
cgp = float(input("Enter your CGPA: "))
val = torch.tensor(cgp)
with torch.no_grad():
    prob = predict(val, w, b)
print("Probability of getting placed: ", prob.item())
```
