```python
# Calculating Grades
```

```python
x = torch.tensor(8.9)
y = torch.tensor(1.0)
```

```python
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
```

```python
def loss_fx(y_pred, y):
    ep = 1e-8
    predict = torch.clamp(y_pred, ep, 1-ep)
    return -(y*torch.log(y_pred) + (1-y)*torch.log(1-y_pred))
```

```python
z = w*x + b
y_pred = torch.sigmoid(z)
loss = loss_fx(y_pred, y)
```

```python
loss.backward()
```

```python
print(w.grad)
print(b.grad)
```
