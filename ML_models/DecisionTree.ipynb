```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier
```

```python
# Scratch code 

# ------------------
# Node
# ------------------
class Node:
    def __init__(self, feature=None, threshold=None, left=None, right=None, value=None):
        self.feature = feature
        self.threshold = threshold
        self.left = left
        self.right = right
        self.value = value


# ------------------
# Entropy
# ------------------
def entropy(y):
    classes = np.unique(y)
    ent = 0
    for c in classes:
        p = np.sum(y == c) / len(y)
        if p > 0:
            ent -= p * np.log2(p)
    return ent


# ------------------
# Information Gain
# ------------------
def information_gain(parent, left, right):
    parent_entropy = entropy(parent)
    n = len(parent)
    weighted_entropy = (len(left)/n * entropy(left)  +  len(right)/n * entropy(right))
    return parent_entropy - weighted_entropy


# ------------------
# Best Split
# ------------------
def best_split(X, y):
    best_gain = -1
    split_feature = None
    split_threshold = None
    n_features = X.shape[1]
    for feature in range(n_features):
        thresholds = np.unique(X[:, feature])
        for threshold in thresholds:
            left_mask = X[:, feature] <= threshold
            right_mask = X[:, feature] > threshold
            if np.sum(left_mask) == 0 or np.sum(right_mask) == 0:
                continue
            gain = information_gain(
                y,
                y[left_mask],
                y[right_mask]
            )
            if gain > best_gain:
                best_gain = gain
                split_feature = feature
                split_threshold = threshold

    return split_feature, split_threshold


# ------------------
# Build Tree
# ------------------
def build_tree(X, y):

    # Pure node
    if len(np.unique(y)) == 1:
        return Node(value=y[0])

    feature, threshold = best_split(X, y)

    if feature is None:
        values, counts = np.unique(y, return_counts=True)
        return Node(value=values[np.argmax(counts)])

    left_mask = X[:, feature] <= threshold
    right_mask = X[:, feature] > threshold

    left_tree = build_tree(
        X[left_mask],
        y[left_mask]
    )

    right_tree = build_tree(
        X[right_mask],
        y[right_mask]
    )

    return Node(
        feature=feature,
        threshold=threshold,
        left=left_tree,
        right=right_tree
    )


# ------------------
# Predict One
# ------------------
def predict_one(x, node):

    if node.value is not None:
        return node.value

    if x[node.feature] <= node.threshold:
        return predict_one(x, node.left)

    return predict_one(x, node.right)


# ------------------
# Predict Multiple
# ------------------
def predict(X, tree):

    predictions = []

    for row in X:
        predictions.append(
            predict_one(row, tree)
        )

    return np.array(predictions)


# ------------------
# Dataset
# ------------------
X = np.array([
    [22,25000],
    [25,30000],
    [35,60000],
    [40,70000],
    [45,80000]
])
y = np.array([0,0,1,1,1])

# ------------------
# Train
# ------------------
tree = build_tree(X, y)

# ------------------
# Test
# ------------------
X_test = np.array([
    [20,20000],
    [38,65000],
    [50,90000]
])
preds = predict(X_test, tree)
print(preds)
```

```python
# --------------------------
# Entropy
# --------------------------
def entropy(y):
    classes, counts = np.unique(y, return_counts=True)
    probs = counts / len(y)
    return -np.sum(probs * np.log2(probs + 1e-10))

# --------------------------
# Information Gain
# --------------------------
def information_gain(X_column, y, threshold):
    parent_entropy = entropy(y)

    left_mask = X_column <= threshold
    right_mask = X_column > threshold

    if sum(left_mask) == 0 or sum(right_mask) == 0:
        return 0

    n = len(y)

    left_entropy = entropy(y[left_mask])
    right_entropy = entropy(y[right_mask])

    child_entropy = (
        len(y[left_mask]) / n * left_entropy
        + len(y[right_mask]) / n * right_entropy
    )

    return parent_entropy - child_entropy

# --------------------------
# Best Split
# --------------------------
def best_split(X, y):
    best_gain = -1
    split_feature = None
    split_threshold = None

    n_features = X.shape[1]

    for feature in range(n_features):
        thresholds = np.unique(X[:, feature])

        for threshold in thresholds:
            gain = information_gain(
                X[:, feature],
                y,
                threshold
            )

            if gain > best_gain:
                best_gain = gain
                split_feature = feature
                split_threshold = threshold

    return split_feature, split_threshold

# --------------------------
# Example Dataset
# --------------------------

X = np.array([[2],[3],[10],[19],[20]])
y = np.array([0,0,1,1,1])
feature, threshold = best_split(X, y)
print("Best Feature :", feature)
print("Best Threshold :", threshold)
```
